# HRSBF — ataque às 24 famílias remanescentes

Os 24 parâmetros de peso três se agrupam em nove classes verificadas por permutações dos índices. O lote usa CPU, reduz apenas equações afins forçadas, audita os modelos originais e verifica cada UNSAT com DRAT-trim. SAT não certifica bentness; timeout mantém a família aberta.

Execute as células em ordem. Nenhuma compra é realizada. Os resultados são salvos por representante; baixe o ZIP ao terminar para preservar os arquivos fora do ambiente temporário.


```python
from pathlib import Path
import os, subprocess, sys
BASE=Path('/content/hrsbf_colab'); BASE.mkdir(exist_ok=True)
PROJECT=BASE/'project'
def command(args, cwd=None):
    subprocess.run([str(v) for v in args],cwd=cwd,check=True)
if not PROJECT.exists():
    command(['git','clone','https://github.com/contatofilipevono-rgb/hrsbf-bent-verification.git',PROJECT])
command(['git','fetch','origin','pesquisa-t32-2026-10-04'],PROJECT)
command(['git','checkout','26ab733406b2fe55f4dd12fe8a2ee855f840df4d'],PROJECT)
for name,url in [('cadical','https://github.com/arminbiere/cadical.git'),('drat-trim','https://github.com/marijnheule/drat-trim.git')]:
    folder=BASE/name
    if not folder.exists():command(['git','clone','--depth','1',url,folder])
    print(name,subprocess.check_output(['git','rev-parse','HEAD'],cwd=folder,text=True).strip())
if not (BASE/'cadical/build/cadical').exists():
    command(['./configure'],BASE/'cadical'); command(['make','-j2'],BASE/'cadical')
if not (BASE/'drat-trim/drat-trim').exists():command(['make','drat-trim'],BASE/'drat-trim')
print('Preparação concluída. CPUs:',os.cpu_count())
```

```python
# CPU; nenhum acelerador GPU é necessário.
# Limites iniciais: duas tarefas, 20 min por representante, lote de até 2 h.
command([sys.executable,PROJECT/'pesquisa_t32/colab_24_reduzidas.py',
         '--project',PROJECT,'--base',BASE,'--workers','2',
         '--seconds','1200','--wall-seconds','7200'])
```
